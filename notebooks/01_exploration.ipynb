```python
## 2. Data Quality Profiling

# Before transforming the healthcare dataset, we first profile the raw
# data to understand its quality.

# The profiling stage checks:

# - Missing values
# - Duplicate records
# - Duplicate encounter identifiers
# - Date validity
# - Length-of-stay consistency
# - Categorical values
# - Numeric distributions

# No records are modified or deleted during this stage. The purpose is
# to identify data-quality issues before defining the cleaning rules.
```

```python
# ### 2.1 Missing Value Analysis

# Missing values are common in healthcare datasets because not every
# measurement or field is available for every encounter.

# We calculate both the number and percentage of missing values for
# each column. This helps determine which fields require validation or
# special handling during the ingestion process.
```

```python
# Calculate missing values for every column
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": (df.isna().mean() * 100).round(2)
})

# Keep only columns that contain missing values
missing = (
    missing[missing["missing_count"] > 0]
    .sort_values("missing_percentage", ascending=False)
)

missing
```

```python
# Display the 20 columns with the highest percentage of missing values
missing.head(20)
```

```python
# ### 2.2 Duplicate Record Analysis

# Duplicate records can cause incorrect analytical results, such as
# inflated encounter counts.

# We therefore check:

# 1. Completely duplicated rows
# 2. Duplicate encounter IDs
# 3. Duplicate patient IDs

# A patient ID is expected to appear multiple times because a patient
# may have multiple healthcare encounters. Therefore, duplicate patient
# IDs are not automatically considered an error.
```

```python
# Count completely duplicated rows
duplicate_rows = df.duplicated().sum()

# Count duplicate encounter IDs
duplicate_encounters = df["encounter_id"].duplicated().sum()

# Count duplicate patient IDs
duplicate_patients = df["patient_id"].duplicated().sum()

print(f"Total records: {len(df):,}")
print(f"Duplicate rows: {duplicate_rows:,}")
print(f"Duplicate encounter IDs: {duplicate_encounters:,}")
print(f"Duplicate patient IDs: {duplicate_patients:,}")


# A patient can legitimately have multiple encounters. Therefore,
# `patient_id` should not be unique in the encounter-level dataset.

# The expected business grain for the future fact table is:

# **One row = one healthcare encounter.**

# Therefore, `encounter_id` is more appropriate as the uniqueness check
# for the encounter fact table.
```

```python
# ### 2.3 Date Validation

# The dataset contains encounter and discharge dates.

# We convert these columns to pandas datetime values using
# `errors="coerce"` so that invalid date values become null instead of
# causing the pipeline to fail unexpectedly.

# We then inspect the number of invalid dates and the overall date range.
```

```python
# Convert date columns to datetime
df["encounter_date"] = pd.to_datetime(
    df["encounter_date"],
    errors="coerce"
)

df["discharge_date"] = pd.to_datetime(
    df["discharge_date"],
    errors="coerce"
)

# Count invalid/missing dates
invalid_encounter_dates = df["encounter_date"].isna().sum()
invalid_discharge_dates = df["discharge_date"].isna().sum()

print(f"Invalid encounter dates: {invalid_encounter_dates:,}")
print(f"Invalid discharge dates: {invalid_discharge_dates:,}")
```

```python
# Inspect the date range
print("Earliest encounter:", df["encounter_date"].min())
print("Latest encounter:", df["encounter_date"].max())

print("Earliest discharge:", df["discharge_date"].min())
print("Latest discharge:", df["discharge_date"].max())
```

```python
# ### 2.4 Encounter and Discharge Date Consistency

# We check whether discharge dates occur before encounter dates.

# A discharge date earlier than the encounter date would indicate a
# potential data-quality issue.

# This check does not modify the source data. It only identifies records
# that require investigation.
```

```python
# Identify records where discharge occurs before the encounter
invalid_date_order = df[
    df["discharge_date"].notna()
    & df["encounter_date"].notna()
    & (df["discharge_date"] < df["encounter_date"])
]

print("Records with discharge before encounter:",
      len(invalid_date_order))
```

```python
# ### 2.5 Length-of-Stay Validation

# The dataset contains a reported `los_days` value as well as encounter
# and discharge dates.

# We calculate an independent length of stay from the two dates and
# compare it with the reported value.

# This provides an example of a cross-field data-quality validation.
```

```python
# Calculate length of stay from the encounter and discharge dates
calculated_los = (
    df["discharge_date"] - df["encounter_date"]
).dt.days

# Compare calculated and reported values
los_comparison = pd.DataFrame({
    "reported_los": df["los_days"],
    "calculated_los": calculated_los
})

los_comparison["difference"] = (
    los_comparison["reported_los"]
    - los_comparison["calculated_los"]
)

los_comparison.head()
```

```python
# Count records where reported and calculated LOS differ
los_mismatch = los_comparison[
    los_comparison["reported_los"].notna()
    & los_comparison["calculated_los"].notna()
    & (los_comparison["difference"] != 0)
]

print("Records with LOS mismatch:", len(los_mismatch))
```

```python
# ### 2.6 Categorical Value Analysis

# Categorical fields should contain standardized values.

# We inspect the unique value distributions of important categorical
# columns before defining accepted values or normalization rules.

# These results will later help us implement data-quality tests in dbt.
```

```python
categorical_columns = [
    "encounter_type",
    "facility_type",
    "note_type",
    "age_band",
    "sex",
    "race_ethnicity",
    "insurance_type",
    "state"
]

for column in categorical_columns:
    print(f"\n{'=' * 60}")
    print(f"{column}")
    print(f"{'=' * 60}")
    print(df[column].value_counts(dropna=False).head(20))
```

```python
# ### 2.7 Numeric Data Profiling

# Numeric healthcare fields are profiled to identify missing values,
# unexpected ranges, and unusual distributions.

# At this stage, the values are only being profiled. We are not making
# clinical conclusions from the measurements.
```

```python
numeric_columns = [
    "age",
    "los_days",
    "n_secondary_dx",
    "n_medications",
    "note_word_count",
    "vs_hr_bpm",
    "vs_sbp_mmhg",
    "vs_dbp_mmhg",
    "vs_rr_breaths_min",
    "vs_temp_f",
    "vs_spo2_pct",
    "vs_weight_kg",
    "vs_height_cm",
    "vs_bmi",
    "lab_glucose",
    "lab_creatinine",
    "lab_wbc",
    "lab_hba1c"
]

numeric_profile = df[numeric_columns].describe().T

numeric_profile
```

```python
# ### 2.8 Data Quality Summary

# The following summary provides a high-level view of the raw dataset,
# including row count, column count, duplicate records, and missing
# values.

# This profile will be used to define the validation and cleaning rules
# implemented in the Python ingestion pipeline.
```

```python
quality_summary = pd.DataFrame({
    "metric": [
        "Total rows",
        "Total columns",
        "Duplicate rows",
        "Duplicate encounter IDs",
        "Duplicate patient IDs",
        "Invalid encounter dates",
        "Invalid discharge dates",
        "Discharge before encounter",
        "Columns with missing values"
    ],
    "value": [
        len(df),
        len(df.columns),
        duplicate_rows,
        duplicate_encounters,
        duplicate_patients,
        invalid_encounter_dates,
        invalid_discharge_dates,
        len(invalid_date_order),
        len(missing)
    ]
})

quality_summary
```

```python
# ### 2.9 Save Data-Quality Profiles

# The profiling results are saved separately from the raw dataset.

# The raw source file remains unchanged, while generated profiling
# artifacts are stored in the processed-data directory.
```

```python
# Save missing-value profile
missing.to_csv(
    "../data/processed/missing_value_profile.csv"
)

# Save numeric profile
numeric_profile.to_csv(
    "../data/processed/numeric_profile.csv"
)

# Save overall quality summary
quality_summary.to_csv(
    "../data/processed/data_quality_summary.csv",
    index=False
)

print("Data-quality profiles saved successfully.")
```
